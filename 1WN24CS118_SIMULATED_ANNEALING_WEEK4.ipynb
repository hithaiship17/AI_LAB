```python

import random
import math

def cost(state):
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing():
    n = 8

    # Random initial state
    state = [random.randint(1, n) for _ in range(n)]
    current_cost = cost(state)

    temperature = 100
    cooling_rate = 0.99

    print("Initial State:", state)
    print("Initial Cost:", current_cost)

    while current_cost != 0 and temperature > 0.01:

        # Generate a random neighboring state
        neighbor = state.copy()
        col = random.randint(0, n - 1)
        row = random.randint(1, n)

        neighbor[col] = row

        new_cost = cost(neighbor)
        delta = new_cost - current_cost

        # Accept better states or probabilistically accept worse states
        if delta < 0 or random.random() < math.exp(-delta / temperature):
            state = neighbor
            current_cost = new_cost

        temperature *= cooling_rate

    print("\nFinal State:", state)
    print("Final Cost:", current_cost)

    if current_cost == 0:
        print("Solution Found!")

        for row in range(1, n + 1):
            for col in range(n):
                if state[col] == row:
                    print("Q", end=" ")
                else:
                    print(".", end=" ")
            print()
    else:
        print("Solution not found. Run again.")


simulated_annealing()
```
